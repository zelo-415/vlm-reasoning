```python
import base64
import os
from openai import OpenAI
from dotenv import load_dotenv

load_dotenv()

api_key = os.getenv("OPENAI_API_KEY")

if not api_key:
    raise ValueError("OPENAI_API_KEY not found in environment variables.") 

client = OpenAI(api_key=api_key)

def encode_image(image_path):
    with open(image_path, "rb") as image_file:
        return base64.b64encode(image_file.read()).decode('utf-8')

def infer_preference(image_path, prompt):
    base64_image = encode_image(image_path)

    response = client.chat.completions.create(
        model = "gpt-4o-mini",
        messages = [
            {
                "role": "system",
                "content": "You are an expert in Human-Robot Interaction (HRI) and can analyze images to determine human preferences. Please provide insights of potential user preferences based on the images provided."
            },
            {
                "role": "user",
                "content": [
                    {"type": "text", "text": prompt},
                    {
                        "type": "image_url",
                        "image_url": {
                            "url": f"data:image/png;base64,{base64_image}",
                            #  "details": "high-resolution"  # Optional: additional details about the image
                        }
                    }
                ]
            }
        ],
        temperature = 0,
    )

    return response.choices[0].message.content

prompt = "Based on the image provided, what are the potential preferences of the user when handling household objects? Consider aspects such as handedness, grip style, and interaction patterns."
print(infer_preference("images/1.png", prompt))
                     
```
