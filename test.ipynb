```python
import base64
import os
from openai import OpenAI
from dotenv import load_dotenv

load_dotenv(override=True)
```

```python
prompt = "Based on the image provided, what are the potential preferences of the user when handling household objects? Consider aspects such as handedness, grip style, and interaction patterns."
```

```python
gpt_api_key = os.getenv("OPENAI_API_KEY")

if not gpt_api_key:
    raise ValueError("OPENAI_API_KEY not found in environment variables.") 

client = OpenAI(api_key=gpt_api_key)

def encode_image(image_path):
    with open(image_path, "rb") as image_file:
        return base64.b64encode(image_file.read()).decode('utf-8')

def infer_preference(image_path, prompt):
    base64_image = encode_image(image_path)

    response = client.chat.completions.create(
        model = "gpt-4o-mini", # gpt-4o-mini or gpt-4o
        messages = [
            {
                "role": "system",
                "content": "You are an expert in Human-Robot Interaction (HRI) and can analyze images to determine human preferences. Please provide insights of potential user preferences based on the images provided."
            },
            {
                "role": "user",
                "content": [
                    {"type": "text", "text": prompt},
                    {
                        "type": "image_url",
                        "image_url": {
                            "url": f"data:image/png;base64,{base64_image}",
                            #  "details": "high-resolution"  # Optional: additional details about the image
                        }
                    }
                ]
            }
        ],
        temperature = 0,
    )

    return response.choices[0].message.content

print(infer_preference("images/1.png", prompt))
                     
```

```python
import anthropic

claude_api_key = os.getenv("CLAUDE_API_KEY")

if not claude_api_key:
    raise ValueError("CLAUDE_API_KEY not found in environment variables.")

client = anthropic.Anthropic(api_key=claude_api_key)

def infer_preference_claude(image_path, prompt):
    base64_image = encode_image(image_path)

    response = client.messages.create(
        model="claude-sonnet-4-5-20250929",  
        max_tokens=1024,
        temperature=0,
        messages=[
            {
                "role": "user",
                "content": [
                    {"type": "text", "text": prompt},
                    {
                        "type": "image",
                        "source": {
                            "type": "base64",
                            "media_type": "image/png",
                            "data": base64_image,
                        }
                    }
                ]
            }
        ]
    )   
    return response.content[0].text

print(infer_preference_claude("images/1.png", prompt))
```

```python
import google.genai as genai
import time

genai_api_key = os.getenv("GENAI_API_KEY")
client = genai.Client(api_key=genai_api_key)

# 1. Upload the video
print("Uploading video...")
my_file = client.files.upload(file="videos/hand_video.mp4")
print(f"Upload complete: {my_file.name}")

# 2. [CRITICAL STEP] Wait for processing to finish
print("Waiting for video processing...")
while True:
    # Retrieve the latest file status
    file_status = client.files.get(name=my_file.name)
    
    if file_status.state == "ACTIVE":
        print("Video is ready!")
        break
    elif file_status.state == "FAILED":
        raise ValueError("Video processing failed.")
    
    print("Still processing...")
    time.sleep(2) # Check every 2 seconds

response = client.models.generate_content(
    model="gemini-3-flash-preview",
    contents=[my_file, prompt],
)

```
